## Comment lire ces notebooks

Chaque cellule qui commence par `!` est une commande du terminal : `!interlis convert-sql ...` s'écrit à l'identique dans une console. Une cellule `%%writefile fichier` écrit son contenu dans un fichier, c'est ainsi que les modèles INTERLIS sont écrits.

Jupyter exécute les cellules dans le dossier du notebook (`notebooks/`). Tous les chemins sont relatifs à ce dossier :

| Dossier | Contenu |
|---|---|
| `../data/`, `../collections/` | les données et les modèles du dépôt (à lire, jamais à modifier) |
| `modeles/` | les modèles officiels téléchargés par `fetch-models` |
| `sorties/` | tout ce que les notebooks écrivent : vos modèles `.ili`, le SQL, le SLD, le JSON-FG... |

`modeles/` et `sorties/` n'existent pas au départ : chaque notebook les crée avec `!mkdir -p modeles sorties`. Ils sont ignorés par git.

Une commande `interlis` renvoie un code de sortie : **0** succès, **2** succès avec remarques (notes ou avertissements, ce n'est pas une erreur), **1** échec. Une cellule `!` n'affiche pas ce code ; pour le voir, ajouter `; echo "code de sortie : $?"` à la fin de la ligne.
